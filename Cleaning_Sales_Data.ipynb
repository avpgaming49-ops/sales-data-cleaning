Importing libraries

```python
import pandas as pd
import numpy as np
```

Reading CSV file

```python
df=pd.read_csv("ecommerce_sales_data.csv")
print(df)
```


Displaying Columns name

```python
print(df.columns.tolist())
```

Remove spaces ,Capitalize columns name and replace "_" to " "

```python
df.columns=df.columns.str.strip().str.capitalize().str.replace("_"," ")
print(df.columns.tolist)
```

Checking duplicate Columns

```python
print(len(df.columns))
print(len(df.columns.unique()))
print(df.info())
```

Analysing dataset-info & sum of null values in dataset

```python
print(df.info())
print(df.isnull().sum())
```

Cleaning column "Order date"

```python
#checking data type of column "Order date"
print(df["Order date"].dtype)
#changing datatype (str to datetime)
df["Order date"] = pd.to_datetime(df["Order date"],errors="coerce")
print("\nFix Applied")
#checking data type
print(df["Order date"].dtype)
#displaying the column
print(df["Order date"])
#checking dataset-info()
print(df.info())
#droping null valeu rows
df=df.dropna(subset="Order date")
#checking sum of null values in column
print(df["Order date"].isnull().sum())
```

Rows sorted ,the small charater word are sorted in last

```python
nf=np.sort(df["Product"].unique())
nf
```

Cleaning Product Column

```python
print(df["Product"].unique())
df["Product"]=df["Product"].str.replace("shoes","Shoes")
print("\nFixed Applied")
print(df["Product"].unique())
```

Cleaning Category Column

```python
df["Category"].unique()
#duplicates with other capital and small letters ae removed
same_map={
    "electronic" : "Electronics",
    "ELECTRONICS" : "Electronics",
    "electronics" : "Electronics",
    "sports" : "Sports"
} 

df["Category"]=df["Category"].str.replace(same_map)
print(df["Category"].unique())

```

```python
# #checking how much % the null values are
print((df["Category"].isna().sum()/len(df))*100)
#filling mode to Category column
df["Category"]=df["Category"].fillna(df["Category"].mode()[0])
print(df["Category"].isnull().sum())
print(df["Category"].dtype)
```

Cleaning Quantity Column

```python
print(df["Quantity"].dtype)
# #coverting string dtype to integer dtype
df["Quantity"]=pd.to_numeric(df["Quantity"],errors="coerce")
# print(df.info())

#counting the nan values in Quantity column
mask=pd.to_numeric(df["Quantity"],errors="coerce").isna()
print(df.loc[mask,["Quantity"]])

#filling nan values to median of column Quantity
df["Quantity"]=df["Quantity"].fillna(df["Quantity"].median())
print(df["Quantity"].dtype)

print(df.isnull().sum())

print(df["Quantity"].describe())
```

Cleaning Price Column

```python
print(df["Price"].dtype)
#converting datatype from str to numeric datatype
df["Price"]=pd.to_numeric(df["Price"],errors="coerce")
#cheching datatype
print(df["Price"].dtype)

#actual length of column Price
print(len(df["Price"]))
#length of column Price grater than zero
print(len(df["Price"]>0))
#display number of notnull values
print(df["Price"].notna().sum())
#display number of null values
print(df["Price"].isna().sum())
#filling median to null values
df["Price"]=df["Price"].fillna(df["Price"].median())

print(df.isnull().sum())
```

Cleaning Total Column

```python
print(df)
#filling null values in column Total with (Quantity * Price)
df["Total"]=df["Total"].fillna(df["Quantity"]*df["Price"])
print(df["Total"])
```


Checking Cleaned Dataset

```python
# checing null values
print(df.isnull().sum())
```

```python
#checing duplicate values
print(df.duplicated().sum())

#cheching duplicate row
df[df.duplicated(keep=False)]

#deleting duplicate column
df=df.drop_duplicates()
print(df.duplicated().sum())
```

Cleaning the negative Rows

```python
print(df["Quantity"].dtype)
# #checking negative values
df[(df["Quantity"]<0) | (df["Price"]<0)]
#removing negative values
df=df[(df["Quantity"]>0) & (df["Price"]>0)]

#checking 
print(df["Quantity"].min())
print(df["Price"].min())
print(df["Total"].min())
print(df["Quantity"]*df["Price"]==df["Total"])
```

Cleaning Column Total

```python
#correcting "Total" column
df["Total"]=df["Quantity"]*df["Price"]
df["Price"]=df["Price"].round(2)
df["Total"]=df["Total"].round(2)

print(df.isnull().sum())
```
